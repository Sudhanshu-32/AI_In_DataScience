**3.Count words,lines,and characters**  
**Problem statement:**  
WAP that reads a text file and displays:  
1.total no of lines  
2.total no of words  
3.total no of characters


```python
# Count words, lines, and characters in 'student.txt'

# Initialize counters for lines, words, and characters
line_count = 0
word_count = 0
char_count = 0

# Open the file in read mode ('r')
# Using 'with open(...)' ensures the file is automatically closed after we're done
with open("student.txt", 'r') as file:
  # Iterate over each line in the file
  for line in file:
    # Increment the line counter for each line read
    line_count += 1

    # Split the line into words using whitespace as a delimiter
    # Then add the number of words in the current line to the total word_count
    words_in_line = line.split()
    word_count += len(words_in_line)

    # Add the number of characters in the current line to the total char_count
    # This includes the newline character at the end of each line
    char_count += len(line)

# Print the final counts
print(f"Total number of lines: {line_count}")
print(f"Total number of words: {word_count}")
print(f"Total number of characters: {char_count}")
```

```python
# Read data from source.txt and copy to destination.txt
try:
  with open('source.txt', 'r') as source_file:
    content = source_file.read()

  with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)

  print("Content successfully copied from source.txt to destination.txt!")

except FileNotFoundError:
  print("Error: source.txt not found. Please make sure the file exists.")
except Exception as e:
  print(f"An error occurred: {e}")
```

**6.Question: Remove Duplicate Lines from a File**  
**Write a Python program to:**  
 Read all lines from a text file.  
 Remove duplicate lines while preserving the original order.  
 Save the unique lines into a new file named unique.txt

```python
with open ("org.txt",'w') as original:
  original.write("python\n")
  original.write("java\n")
  original.write("python\n")
  original.write("c++\n")
  original.write("java\n")
```

### Removing Duplicate Lines Using a List and a Set

This method involves reading each line from the source file. A `set` is used to keep track of lines that have already been encountered, allowing for fast duplicate checking. A `list` is used to store the unique lines in their original order of appearance. Finally, these unique lines are written to a new file.

```python
# Initialize an empty list to store unique lines while preserving order
unique_lines_list = []

# Initialize an empty set to quickly check for seen lines (sets have O(1) average time complexity for lookups)
seen_lines_set = set()

try:
  # Step 1: Read all lines from the source file ('org.txt')
  with open('org.txt', 'r') as source_file:
    for line in source_file:
      # It's important to strip whitespace, including newline characters, for accurate comparison.
      # If not stripped, 'python\n' and 'python' would be treated as different.
      stripped_line = line.strip()

      # Step 2: Check if the stripped line has already been seen
      if stripped_line not in seen_lines_set:
        # If it's a new unique line:
        # Add it to the set of seen lines
        seen_lines_set.add(stripped_line)
        # Add the original line (including its newline character) to our ordered list
        unique_lines_list.append(line)

  # Step 3: Write the unique lines (preserving their original order) to a new file
  with open('unique_by_list_set.txt', 'w') as destination_file:
    for unique_line in unique_lines_list:
      destination_file.write(unique_line)

  print("Successfully removed duplicate lines and saved unique content to 'unique_by_list_set.txt'!")

except FileNotFoundError:
  print("Error: 'org.txt' not found. Please ensure the file exists.")
except Exception as e:
  print(f"An unexpected error occurred: {e}")
```

To confirm the content of the newly created `unique_by_list_set.txt` file, you can read and display it:

```python
# Read and display the content of 'unique_by_list_set.txt'
with open('unique_by_list_set.txt', 'r') as file:
  print("Content of unique_by_list_set.txt:")
  print(file.read())
```
