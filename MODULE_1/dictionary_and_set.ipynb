#**Dictionary and set questions**

**1. Write a program to create a dictionary from two lists: one of keys and one of values.**

```python
# Write a program to create a dictionary from two lists: one of keys and one of values.
keys = ['name', 'age', 'city']
values = ['Alice', 30, 'New York']
# The zip() function in Python is used to combine multiple iterables (like lists, tuples, etc.)
# into a single iterable of tuples. It pairs up elements
# from each of the input iterables based on their index.
my_dictionary = dict(zip(keys, values))
print(my_dictionary)
```

**2. Merge two dictionaries into one**

```python
# Merge two dictionaries into one
dict1={"name":"sudhanshu","age":19}
dict2={"branch":"IT","course":"btech"}
dict3=dict1 | dict2
print(dict3)
```

**4.Given two sets ,check if one set is a subset of another.**

```python
# Given two sets ,check if one set is a subset of another.
set1={2,4,5,"python","a",6}
set2={4,5,"a"}
result=set2.issubset(set1)
print(result)
```

**5.Write a program to check whether two list have at leastone common element using sets.**

```python
# Write a program to check whether two list have at least one common element using sets.
list1=[2,4,5,7,9]
list2=[5,6,78,3]
set1=set(list1)
set2=set(list2)
common= set1.intersection(set2)
if bool(common):
  print("there are common element",common)
else:
  print("no common element")
```
