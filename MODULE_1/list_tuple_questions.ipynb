#**List and Tuple questions**

**1.Write a program to find the largest and smallest elements in a list**

```python
# Write a program to find the largest and smallest elements in a list
list1=[4,5,6,89,1,34]
list1.sort()
print(f"largest element is {list1[-1]}")
print(f"smallest element is {list1[0]}")
```

**2.Write a program to remove duplicate elements from a list.**

```python
# Write a program to remove duplicate elements from a list.
list1=[2,2,4,5,6,9,20,20,30,9,5]
# first convert list to set to remove duplocates and then change it to list
temp=set(list1)
list1=list(temp)
print(list1)
```

**3.Write a program to reverse a list without using built-in reverse functions**

```python
# Write a program to reverse a list without using built-in reverse functions.
list1=[23,45,67,89]
# using slicing method
print(list1[::-1])
```

**4. Write a program to count even and odd numbers in a list.**

```python
# Write a program to count even and odd numbers in a list.
list1=[2,4,5,6,8,9,12]
odd=[i for i in list1 if i%2!=0]
print(f"there are {len(odd)} odd numbers in list")
even=[j for j in list1 if j%2==0]
print(f"there are {len(even)} even numbers in list")
```

**5. Write a program to merge two lists and sort the final list.**

```python
# Write a program to merge two lists and sort the final list.
l1=[3,4,6,7]
l2=[4,5,7,8]
l3=l1+l2
l3.sort()
print(l3)
```

**7.Write a program to check whether an element exists in a tuple.**

```python
# Write a program to check whether an element exists in a tuple.
t1=(5,7,9,8,23,45,75)
element=int(input("enter the element to search:"))
print(f"{element} is at index {t1.index(element)}.")
```

**8. Write a program to count the occurrence of an element in a tuple**
