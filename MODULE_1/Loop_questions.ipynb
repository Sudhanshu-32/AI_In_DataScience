**1.Write a Python program to print multiplication tables from 1 up to an entered number side-by-side using nested while loops only.**

```python
#Write a Python program to print multiplication tables from 1 up to an entered number side-by-side using nested while loops only.
n=int(input("enter the number upto which want to print table:"))
i=1
while i<=10:
  j=1
  while j<=n:
    res=i*j
    print(f"{i} x {j} = {res}", end="\t")
    j+=1
  print()
  i+=1
```

```python
# using for loop
n = int(input("enter the number upto which want to print table:"))

for i in range(1, 11):  # Outer loop for multiplier (1 to 10)
  for j in range(1, n + 1): # Inner loop for table number (1 to n)
    res = i * j
    print(f"{i} x {j} = {res}", end="\t")
  print() # Move to the next line after each row of tables
```

**2.Reverse the no without using reverse or slicing methods.**

```python
#2.Reverse the no without using reverse or slicing methods.
num=int(input("Enter the number:"))
rev=0
print(f"original number is :{num}")
while num>0:
  digit=num%10
  rev= rev*10+digit
  num=num//10

print(f"reversed number is :{rev}")
```

**3.Check number is palindrome or not**

```python
# Check number is palindrome or not
num=int(input("Enter the number:"))
rev=0
n=num
while num>0:
  digit=num%10
  rev= rev*10+digit
  num=num//10
if rev==n:
  print("number is palindrome")
else:
  print("number is not palindrome")
```

**6.find the largest digit in number**

```python
# find the largest digit in number
num=int(input("enter the number:"))
n=num
l1=[]
while num>0:
  digit=num%10
  num=num//10
  l1.append(digit)
  l1.sort()
max=l1[-1]

print(f"the largest digit is {max}")
```

```python
# Alternative method to find the largest digit in a number
num = int(input("Enter the number:"))

max_digit = 0 # Initialize max_digit to 0
temp_num = num # Use a temporary variable to extract digits

# Handle the case of 0 separately, as the loop won't run for 0
if temp_num == 0:
    max_digit = 0

while temp_num > 0:
  digit = temp_num % 10 # Get the last digit
  if digit > max_digit:
    max_digit = digit # Update max_digit if current digit is larger
  temp_num //= 10 # Remove the last digit

print(f"The largest digit is {max_digit}")
```
