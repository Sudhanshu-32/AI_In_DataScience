 **1.Write a program to check whether a year is a leap year or not.**

```python
# 1. Write a program to check whether a year is a leap year or not.
year=int(input("enter the year:"))
if year % 4 ==0:
  if year % 100 !=0:
    print(f"{year} is a leap year")
elif year % 400 ==0:
  print(f"{year} is a leap year")
else:
  print(f"{year} is not a leap year")

# or
# -> in just one line
# if (year%400==0) or {year%4==0 and year%100!=0}:
# print("leap year")

```

**2.Write a program to find the largest among three numbers using nested conditional statements.**

```python
# Write a program to find the largest among three numbers using nested conditional statements.
n1=int(input("enter first number:"))
n2=int(input("enter second number:"))
n3=int(input("enter third number:"))
if n1>n2:
    if n1>n3:
      print(f"{n1} is the largest number")

elif n2>n3:
  if n2>n1:
    print(f"{n2} is the largest number")
else:
  print(f"{n3} is the largest number")
```

**3. Write a program to check whether a character is an uppercase letter, lowercase letter, digit, or
special character.**

```python
#  Write a program to check whether a character is an uppercase letter, lowercase letter,
#  digit, or special character
char=input("enter the characters:")
if char.isupper()==True:
  print(f"{char} has all uppercase letters.")
elif char.islower()==True:
  print(f"{char} has all lowercase letters.")
elif char.isdigit()==True:
  print(f"{char} has all digit:")
else:
  print(f"{char} is special character.")
```

**5.Write a program to determine whether a triangle is Equilateral, Isosceles, or Scalene.**

```python
# Write a program to determine whether a triangle is Equilateral, Isosceles, or Scalene.
a=int(input("enter the side a:"))
b=int(input("enter the side b:"))
c=int(input("enter the side c:"))
if a==b==c:
  print("This is an equilateral triangle")
elif (a==b!=c) or (a!=b==c):
  print("this is an Iscosceles triangle")
else:
  print("this is a scalene triangle")
```

**6. Write a program to create a simple calculator using if-elif-else.**


```python

#  Write a program to create a simple calculator using if-elif-else.
a=float(input("enter first number:"))
b=float(input("enter second number:"))
operation=input("enter the operation to perform:")
if operation=="+":
  sum=a+b
  print(f"sum of a and b is {sum}")
elif operation=="-":
  sub=a-b
  print(f"difference of a and b is {sub}")
elif operation=="*":
  mul=a*b
  print(f"multiple of a and b is {mul}")
elif operation=="/":
  if b==0:
    print("error div by zero")
  else:
   div=a/b
   print(f"a divided by b gives:{div}")
elif operation=="%":
  if b==0:
    print("error div by zero")
  else:
    mod=a%b
    print(f"a modulo b is {mod}")
else:
  print("enter valid operation")

```
