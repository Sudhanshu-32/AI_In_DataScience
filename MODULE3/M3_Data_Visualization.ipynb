#**3.2 Histogram**

```python
# Import the matplotlib library for plotting
import matplotlib.pyplot as plt

# Define a list of ages
age=[20,22,22,23,24,24,25,22,22,23,25]

# Create a histogram of the age distribution
plt.hist(age)

# Add a title to the histogram
plt.title("Age distribution")

# Label the x-axis
plt.xlabel("Age",fontsize=16)

# Label the y-axis
plt.ylabel("Distribution",fontsize=16)

# Display the plot
plt.show()
```

```python
# Import the matplotlib library for plotting
import matplotlib.pyplot as plt

# Define a list of ages
age=[20,22,22,23,24,24,25,22,22,23,25]

# Create a histogram of the age distribution with custom settings:
# bins=5: Divides the data into 5 intervals (bars)
# edgecolor="r": Sets the border color of bars to red
# color="skyblue": Sets the fill color of bars to sky blue
# alpha=1: Sets the transparency (1 means fully opaque)
plt.hist(age,bins=5,edgecolor="r",color="skyblue",alpha=1)

# Add a title to the histogram
plt.title("Age distribution")

# Label the x-axis
plt.xlabel("Age",fontsize=16)

# Label the y-axis
plt.ylabel("Distribution",fontsize=16)

# Add a grid to the plot, specifically for the y-axis
plt.grid(axis="y")

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = [22, 24, 19, 25, 30, 28, 21, 35, 40, 18]

median_val = np.median(data)

plt.hist(data, bins=5, edgecolor='black')
plt.axvline(median_val, color='red', linestyle='--', linewidth=2, label=f'Median: {median_val}')

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

**3.3 BAR CHART**

```python
# bar chart
# defining dataset subject and score

subjects=["python","c","java","phy","chem"]

score=[70,75,80,95,90]

# plotting the bar chart and adding color

plt.bar(subjects,score,color=["red","green","yellow","blue"])

# adding title and labels

plt.title("Bar chart of subject and score")

plt.xlabel(subjects)

plt.ylabel(score)

# adding grid

plt.grid(axis="y")

# showing the chart

plt.show()
```

**3.4 PIE CHART**

```python
# PIE chart
# defining dataset subject and score

subjects=["python","c","java","phy","chem"]

score=[70,75,80,95,90]

colors=["green","yellow","skyblue","black","gray"]

# plotting the PIE chart and adding percentage ,startangle

plt.pie(score,labels=subjects,autopct="%1.2f%%",startangle=(0-360),explode=(0,0,0,0,0.5),colors=["green","yellow","skyblue","black","gray"],shadow=True)

# adding title and labels

plt.title("pie chart of subject and score")


# adding legend
# plt.legend(loc="upper_right")
# plt.legend(loc="best")
# plt.legend(loc="")


# showing the chart

plt.show()
```

**3.5 SCATTER PLOT**

```python
# scatter plot
# defining dataset

age=[23,24,25,30,87]

scores=[20,30,40,35,32]



rang=["green","yellow","skyblue","black","gray"]

size=[50,500,800,600,200]

plt.scatter(age,scores,c=rang,s=size,alpha=0.7)

plt.colorbar(label="intensity")


# adding title and labels

plt.title("scatter plot of age and score")

plt.xlabel(age)

plt.ylabel(scores)



# showing the chart

plt.show()
```

```python
# adding multiple categories
# scatter plot
# defining dataset

age=[23,24,25,30,87]

scores=[20,30,40,35,32]


age1=[30,35,45,60,117]

scores1=[30,50,70,80,90]



rang=["green","yellow","skyblue","blue","pink"]

size=[50,500,800,600,200]

plt.scatter(age,scores,c=rang,s=size,alpha=0.9)

plt.scatter(age1,scores1,c=rang,s=size,alpha=0.5)

plt.colorbar(label="intensity")


# adding title and labels

plt.title("scatter plot of age and score")

plt.xlabel(age)

plt.ylabel(scores)


plt.legend(loc="upper right")

# showing the chart

plt.show()
```

**Task 1**  
Plot the subject and score of three different years of class 10th student

```python
import numpy as np
import matplotlib.pyplot as plt

subjects = ["science", "hindi", "english", "maths", "sst"]
score_2026 = [90, 92, 95, 99, 98]
score_2025 = [70, 75, 82, 85, 90]
score_2024 = [60, 70, 80, 50, 85]

x = np.arange(len(subjects))   # [0,1,2,3,4]
width = 0.25                    # width of each bar

plt.bar(x - width, score_2024, width, color="red", label="2024")
plt.bar(x, score_2025, width, color="yellow", label="2025")
plt.bar(x + width, score_2026, width, color="green", label="2026")

plt.title("Bar chart of subject and marks of class 10th")
plt.xlabel("----subjects----")
plt.ylabel("marks")

plt.xticks(x, subjects)   # replace numeric ticks with subject names
plt.legend(loc="upper left")
plt.show()
```

Take three diffrent dataset with month,unit and profit.plot all 5 plots

```python
# defining dataset

month=["january","february","march","april","may","june","july","august","september","october","november","december"]
unit=[100,200,300,410,520,600,700,850,900,1000,1140,1210]
profit=[100000,200000,300000,380000,480000,600000,700000,810000,900000,1000000,1100000,1200000]
```

```python
# LINE CHART
import matplotlib.pyplot as plt

plt.figure(figsize=(12,12))
# Plot the data as a line graph
plt.plot(month,profit,color="green",marker="o", label="Profit")

# Add a title to the plot
plt.title("month and profit relation")

# Label the x-axis
plt.xlabel("month")

# Label the y-axis
plt.ylabel("profit")

plt.grid()



for i in range(len(month)):
  plt.text(month[i],profit[i],str([month[i],profit[i]]),va="baseline")

plt.legend(loc="upper left")
# Display the plot
plt.show()
```

```python
# Histogram
import numpy as np

median_val = np.median(unit)

plt.hist(unit, bins=12, edgecolor='black')
plt.axvline(median_val, color='red', linestyle='--', linewidth=2, label=f'Median: {median_val}')

plt.xlabel("unit")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
# Bar chart
import matplotlib.pyplot as plt

plt.figure(figsize=(12,12))
plt.bar(month,unit, label="Units")

# adding title and labels

plt.title("Bar chart of month and unit")

plt.xlabel("Month")

plt.ylabel("Unit")

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# adding grid

plt.grid(axis="y")

plt.legend()

# showing the chart

plt.show()
```

```python
# pie chart
Q_1=month[0:3]
Q_2=month[3:6]
Q_3=month[6:9]
Q_4=month[9:]
quarter=["Q1","Q2","Q3","Q4"]
Q_profit=[sum(profit[0:3]),sum(profit[3:6]),sum(profit[6:9]),sum(profit[9:12])]
```
